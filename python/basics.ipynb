Arithmetic Operators & Comparison Operators

```python
x = 10; y = 3

sum = x + y # summation
sub = x - y # subtraction
mult = x * y # multiplication
div = round(x / y, 2) # division

rem = x % y # remainder

print(sum, sub, mult, div, rem)
```

```python
x = 10; y = 3

greater_than = x > y

greater_than_equal = (10 >= 10)

print(greater_than, greater_than_equal)

less_than = x < y

less_than_equal = (10 <= 10)

print(less_than, less_than_equal)

equal = (x == y)

print(equal)

not_equal = (x != y)

print(not_equal)
```

Precedence and Associativity

```python
eq = 10 + 10 / 2 - 5 * 2 
# higher precedence: / *
# lower precedence: + -
# if eq has multiple higher precedence, then associativity is happened (left to right)
eq

# higher precedence to lower precedence:
# 1. == (Comparison)
# 2. not (Logical NOT)
# 3. and (Logical AND)
# 4. or (Logical OR) 
```
