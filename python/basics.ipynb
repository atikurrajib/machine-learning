Arithmetic Operators & Comparison Operators

```python
x = 10; y = 3

sum = x + y # summation
sub = x - y # subtraction
mult = x * y # multiplication
div = round(x / y, 2) # division

rem = x % y # remainder

print(sum, sub, mult, div, rem)
```

```python
x = 10; y = 3

greater_than = x > y

greater_than_equal = (10 >= 10)

print(greater_than, greater_than_equal)

less_than = x < y

less_than_equal = (10 <= 10)

print(less_than, less_than_equal)

equal = (x == y)

print(equal)

not_equal = (x != y)

print(not_equal)
```

```python
eq = 10 + 10 / 2 - 5 * 2 
# higher precedence: / *
# lower precedence: + -
# if eq has multiple higher precedence, then associativity is happened (left to right)
eq

# higher precedence to lower precedence:
# 1. == (Comparison)
# 2. not (Logical NOT)
# 3. and (Logical AND)
# 4. or (Logical OR) 
```

Break and Continue Statements

```python
sum = 0

for i in range(1, 11):

    if i % 2 == 0:
        continue
    sum += i
print(sum)
```

```python
sum = 0

for i in range(1, 10):
    if i % 4 == 0:
        break
    sum += i
print(sum)
```

```python
accuracy = 95

for i in range(20):
    accuracy += 1
    print(accuracy)
    if accuracy == 100:
        break
```

```python
accuracy = 95

while True:
    if accuracy == 100:
        break
    accuracy += 1
print(accuracy)
```

Problem Solving

```python
# https://codeforces.com/group/MWSDmqGsZm/contest/219158/problem/K
inp = input()

numbers = inp.split()

x = int(numbers[0])
y = int(numbers[1])
z = int(numbers[2])

min = x
max = x

# min
if min > y:
    min = y
if min > z:
    min = z

# max
if max < y:
    max = y
if max < z:
    max = z

print(min, max)
```

```python
# https://codeforces.com/group/MWSDmqGsZm/contest/219432/problem/C
n = int(input())

inp = input()
numbers = inp.split()

even = 0; odd = 0; positive = 0; negative = 0

for i in range(n):

    x = int(numbers[i])
    # positive or negative
    if x > 0:
        positive += 1
    elif x < 0:
        negative += 1
    # odd or even
    if x % 2 != 0:
        odd += 1
    else:
        even += 1

print("Even: ", even)
print("Odd: ", odd)
print("Positive: ", positive)
print("Negative: ", negative)



```

```python
# https://codeforces.com/group/MWSDmqGsZm/contest/219432/problem/Q
t = int(input())

for i in range(t):
    number = int(input())
    if number == 0:
        print(0)
        continue
    while number > 0:
        print(number%10, end=" ")
        number //= 10
    print()

```
